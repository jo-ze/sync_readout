```python
%matplotlib widget
import asyncio
import numpy as np
import matplotlib.pyplot as plt

rows, cols = 10, 8
fig, ax = plt.subplots(figsize=(7, 4))
im = ax.pcolormesh(np.arange(cols), np.arange(rows), np.zeros((rows, cols)),
                    shading='nearest', cmap='RdBu_r')
cbar = fig.colorbar(im, ax=ax, label='test value')
ax.set_xlabel('column')
ax.set_ylabel('row')
plt.show()

async def clim_test(n_iter=60, wait=0.15, reshape_every=15):
    global im, cols
    for i in range(n_iter):
        if i > 0 and i % reshape_every == 0:
            cols = cols + 3 if cols < 20 else 8
            ax.clear()
            M = np.random.rand(rows, cols) * (i + 1) * 5
            im = ax.pcolormesh(np.arange(cols), np.arange(rows), M,
                                shading='nearest', cmap='RdBu_r')
            ax.set_xlabel('column')
            ax.set_ylabel('row')
        else:
            M = np.random.rand(rows, cols) * (i + 1) * 5   # Wertebereich waechst absichtlich
            im.set_array(M.ravel())
            im.set_clim(M.min(), M.max())

        cbar.update_normal(im)
        ax.set_title(f'iter {i}')
        fig.canvas.draw_idle()

        print(f'i={i:3d} cols={cols:2d}  im.clim=({im.get_clim()[0]:7.1f}, {im.get_clim()[1]:7.1f})  '
              f'cbar.ticks=({cbar.get_ticks()[0]:.1f} .. {cbar.get_ticks()[-1]:.1f})')

        await asyncio.sleep(wait)

await clim_test()
```
